```python
import os
import shutil
import pandas as pd
import random
```

```python
# Define paths
source_dir = '/mnt/nis_lab_research/pp_pkg/results/runs/r4/meta_clusters/pp_crawler_baseline'
output_dir = '/mnt/nis_lab_research/pp_pkg/results/runs/master/macro_mcs/se/all_clean_4_test/benign'
csv_file = '/mnt/nis_lab_research/pp_pkg/results/runs/r4/final/pp_crawler_baseline/check.csv'
prefix = 'r4_bl'

NUM_SAMPLES = 14
RANDOM_SEED = 123
random.seed(RANDOM_SEED)
```

```python
# Create the output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)
```

```python
# Read the CSV file and extract the clusters column
df = pd.read_csv(csv_file)
clusters = df['clusters'].tolist()
```

```python
len(clusters)
```

```python
# List all subdirectories in the source directory
subdirs = [d for d in os.listdir(source_dir) if os.path.isdir(os.path.join(source_dir, d))]
```

```python
len(subdirs)
```

```python
# Filter subdirectories that match the clusters in the CSV
filtered_subdirs = [subdir for subdir in subdirs if subdir not in clusters]
```

```python
# Randomly select subdirectories
num_samples_to_select = min(NUM_SAMPLES, len(filtered_subdirs))  # Ensure we don't select more than available
random_subdirs = random.sample(filtered_subdirs, num_samples_to_select)
```

```python
# Copy selected subdirectories to the output directory
for subdir in random_subdirs:
    source_path = os.path.join(source_dir, subdir)
    dest_path = os.path.join(output_dir, f'{prefix}_{subdir}')
    shutil.copytree(source_path, dest_path)
    print(f"Copied {subdir} to {dest_path}")
```
