```python
import os
import pandas as pd
import shutil
```

```python
base_dir = './runs/r5/final'
```

```python
def create_cln_mc_oi(base_path, to_name, lst):
    for mc in lst:
        mc_path_from = os.path.join(base_path, "mc_oi", mc)
        mc_path_to = os.path.join(base_path, f"mc_oi_cln_{to_name}", mc)
        shutil.copytree(mc_path_from, mc_path_to, dirs_exist_ok=True)
```

```python
def get_tot_num_clicks(path):
    with open(path, 'r') as file:
        for line in reversed(file.readlines()):
            last_line = line
            break
    return str(int((last_line.split(':')[-1])))
```

```python

for crawler in os.listdir(base_dir):
    base_path = os.path.join(base_dir, crawler)
    
    df = pd.read_csv(os.path.join(base_path, "check_cln.csv"))
    
    dir_name_list = ['sus', 'se']
    final_info_text_list = ['Suspicious', 'SE']
    mc_lists = [list(df[df['suspicious'] == 1]['clusters']), list(df[df['se'] == 1]['clusters'])]

    with open(os.path.join(base_path, 'final_info.txt'), 'w+') as f:
        tot_pages = 0
        for i, name in enumerate(dir_name_list):
            create_cln_mc_oi(base_path, name, mc_lists[i])
            f.write(f'{final_info_text_list[i]} Clusters: {len(mc_lists[i])}\n')
            tot_pages += len(mc_lists[i])
        f.write(f'Total: {tot_pages}\n\n')
        tot_num_clicks = get_tot_num_clicks(os.path.join(base_path, 'tot_num_clicks.txt'))
        f.write(f'Total Clicks: {tot_num_clicks}')
```
